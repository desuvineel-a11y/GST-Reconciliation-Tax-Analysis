```python
# Sales GST summary
sales_gst = sales_df[["cgst", "sgst", "igst"]].sum()

output_gst = sales_gst.sum()

# Purchase ITC summary
eligible_purchases = purchases_df[
    purchases_df["itc_claimed"].astype(str).str.upper().str.strip() == "Y"
]

eligible_itc = eligible_purchases["gst_amount"].sum()

# Net GST payable
net_payable = output_gst - eligible_itc

print("OUTPUT GST SUMMARY")
print(sales_gst)
print("Total output GST:", round(output_gst, 2))

print("\nINPUT TAX CREDIT")
print("Eligible purchase records:", len(eligible_purchases))
print("Eligible ITC:", round(eligible_itc, 2))

print("\nNET GST PAYABLE")
print(round(net_payable, 2))
```

```python
sales_df["expected_total"] = (
    sales_df["taxable_value"]
    + sales_df["cgst"]
    + sales_df["sgst"]
    + sales_df["igst"]
)

sales_df["total_difference"] = (
    sales_df["total_amount"] - sales_df["expected_total"]
).round(2)

tax_mismatches = sales_df[
    sales_df["total_difference"].abs() > 0.01
]

print("Total sales rows:", len(sales_df))
print("Rows with total amount mismatches:", len(tax_mismatches))
print("\nSample mismatches:")
display(
    tax_mismatches[
        ["invoice_no", "taxable_value", "cgst", "sgst",
         "igst", "total_amount", "expected_total",
         "total_difference"]
    ].head(10)
)
```

```python
print("Purchase rows:", len(purchases_df))

print("\nITC claimed values:")
print(purchases_df["itc_claimed"].value_counts(dropna=False))

print("\nGST amount total:",
      round(purchases_df["gst_amount"].sum(), 2))

print("\nITC claimed amount:",
      round(eligible_purchases["gst_amount"].sum(), 2))

print("\nITC not claimed amount:",
      round(
          purchases_df.loc[
              purchases_df["itc_claimed"].astype(str).str.upper().str.strip() != "Y",
              "gst_amount"
          ].sum(),
          2
      ))
```

```python
net_payable = output_gst - eligible_itc

print("Total Output GST:", round(output_gst, 2))
print("Eligible ITC marked Y:", round(eligible_itc, 2))
print("Preliminary Net GST Payable:", round(net_payable, 2))
```

```python
import pandas as pd
import os

# Create output folder
output_folder = "../outputs"
os.makedirs(output_folder, exist_ok=True)

file_path = os.path.join(
    output_folder,
    "GST_Reconciliation_Summary.xlsx"
)

# 1. GST Summary
gst_summary = pd.DataFrame({
    "Metric": [
        "Total Taxable Sales",
        "CGST Collected",
        "SGST Collected",
        "IGST Collected",
        "Total Output GST"
    ],
    "Amount": [
        sales_df["taxable_value"].sum(),
        sales_df["cgst"].sum(),
        sales_df["sgst"].sum(),
        sales_df["igst"].sum(),
        output_gst
    ]
})

# 2. ITC Summary
itc_summary = pd.DataFrame({
    "Metric": [
        "Total Purchase GST",
        "ITC Marked Y",
        "ITC Marked N",
        "Eligible Purchase Records",
        "Non-claimed Purchase Records"
    ],
    "Value": [
        purchases_df["gst_amount"].sum(),
        eligible_itc,
        purchases_df.loc[
            purchases_df["itc_claimed"].astype(str).str.upper().str.strip() != "Y",
            "gst_amount"
        ].sum(),
        len(eligible_purchases),
        len(purchases_df) - len(eligible_purchases)
    ]
})

# 3. Net Payable
net_summary = pd.DataFrame({
    "Metric": [
        "Total Output GST",
        "ITC Marked Y",
        "Preliminary Net GST Payable"
    ],
    "Amount": [
        output_gst,
        eligible_itc,
        net_payable
    ]
})

# 4. Tax Errors
tax_errors = tax_mismatches[
    [
        "invoice_no", "taxable_value", "cgst", "sgst",
        "igst", "total_amount", "expected_total",
        "total_difference"
    ]
].copy()

with pd.ExcelWriter(file_path, engine="openpyxl") as writer:
    gst_summary.to_excel(writer, sheet_name="GST Summary", index=False)
    itc_summary.to_excel(writer, sheet_name="ITC Summary", index=False)
    net_summary.to_excel(writer, sheet_name="Net Payable", index=False)
    tax_errors.to_excel(writer, sheet_name="Tax Errors", index=False)

print("Excel workbook created successfully!")
print("Saved at:", os.path.abspath(file_path))
```
